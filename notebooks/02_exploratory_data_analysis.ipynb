## Numerical Variable Analysis

This section examines the distribution of the numerical variables within the dataset. Histograms provide insight into the shape of each distribution, while boxplots help identify potential outliers.

```python
#Numerical Variables
numerical_columns = [
    "Age",
    "RestingBP",
    "Cholesterol",
    "MaxHR",
    "Oldpeak",
]

numerical_columns
```

```python
#Histograms
for column in numerical_columns:
    plot_histogram(
        dataframe=df,
        column=column,
        output_path=FIGURES_DIR / f"{column.lower()}_histogram.png",
    )
```

```python
#Boxplots
for column in numerical_columns:
    plot_boxplot(
        dataframe=df,
        column=column,
        output_path=FIGURES_DIR / f"{column.lower()}_boxplot.png",
    )
```

## Interpretation of Numerical Variables

The histograms and boxplots provide an overview of the distribution and spread of the numerical variables in the dataset.

Key observations include:

- Distribution shape (e.g., symmetric or skewed)
- Presence of potential outliers
- Range and variability of measurements

These observations provide important context for subsequent modelling and governance analyses.

## Categorical Variable Analysis

This section examines the frequency distributions of the categorical variables used throughout the empirical validation pipeline.

```python
#Categorical Variables
categorical_columns = [
    "Sex",
    "ChestPainType",
    "RestingECG",
    "ExerciseAngina",
    "ST_Slope",
]

categorical_columns
```

```python
#Count Plots
for column in categorical_columns:
    plot_countplot(
        dataframe=df,
        column=column,
        output_path=FIGURES_DIR / f"{column.lower()}_countplot.png",
    )
```
